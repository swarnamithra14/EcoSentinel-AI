### Objectives:
1. Load trained YOLO11n checkpoint (`models/waste_yolo11n.pt`).
2. Run standalone inference on test split images.
3. Visualize bounding boxes, predicted classes, confidence scores, and detection counts.
4. Validate inference latency and ensure zero retraining during prediction.

```python
import os
import sys
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

sys.path.append('..')
from src.waste.predict import predict_waste

test_dir = Path('../data/processed/waste/images/test')
test_images = sorted(list(test_dir.glob('*.*')))
print(f"Found {len(test_images)} test images for evaluation.")
```

## 1. Run Prediction on Held-Out Test Image

```python
if test_images:
    sample_img = test_images[0]
    result = predict_waste(sample_img, conf_threshold=0.003, model_path='../models/waste_yolo11n.pt')
    print(f"Status: {result['status']}")
    print(f"Total Detections: {result['total_detections']}")
    print(f"Class Counts: {result['counts']}")
    
    if result['annotated_image']:
        plt.figure(figsize=(10, 8))
        plt.imshow(result['annotated_image'])
        plt.axis('off')
        plt.title(f"Waste Detections: {sample_img.name}", fontsize=14)
        plt.show()
else:
    print("No test images found.")
```
